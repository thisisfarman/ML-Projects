```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
```

```python
from sklearn.impute import SimpleImputer
```

```python
df=pd.read_csv("loan_approval_data.csv")
```

```python
df
df.info()
```

```python
df.describe()
```

```python
df.isnull().sum()
```

```python
num_i=SimpleImputer(strategy='mean')
```

```python
numerical_columns=df.select_dtypes(include=["number"]).columns
```

```python
numerical_columns
```

```python
df[numerical_columns]=num_i.fit_transform(df[numerical_columns])
```

```python
categorical_columns=df.select_dtypes(include=["object"]).columns
```

```python
obj_i=SimpleImputer(strategy='most_frequent')
df[categorical_columns]=obj_i.fit_transform(df[categorical_columns])
```

```python
df.info()
```

```python
#EDA
```

```python
from sklearn.preprocessing import LabelEncoder,OneHotEncoder
le=LabelEncoder()
df["Education_Level"]=le.fit_transform(df["Education_Level"])
df["Loan_Approved"]=le.fit_transform(df["Loan_Approved"])
```

```python
cols=["Employment_Status","Marital_Status","Loan_Purpose","Property_Area","Gender","Employer_Category"]
```

```python
ohe=OneHotEncoder(drop="first",sparse_output=False,handle_unknown="ignore")
```

```python
e=ohe.fit_transform(df[cols])
```

```python
enco=pd.DataFrame(e,columns=ohe.get_feature_names_out(cols),index=df.index)
```

```python
df=pd.concat([df.drop(columns=cols),enco],axis=1)
```

```python
df.head()
```

```python
df.info()
```

```python
x=df.drop(columns=["Loan_Approved"],axis=1)
y=df["Loan_Approved"]
from sklearn.preprocessing import StandardScaler
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
ss=StandardScaler()
x_scaled=ss.fit_transform(x_train)
x_test_scaled=ss.transform(x_test)

from sklearn.linear_model import LogisticRegression
lr=LogisticRegression()
lr.fit(x_scaled,y_train)
y_pred=lr.predict(x_test_scaled)
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, confusion_matrix
print("logistic regression")
print("precison :",precision_score(y_test,y_pred))
print("accuracy :",accuracy_score(y_test,y_pred))
print("recall :", recall_score(y_test,y_pred))
print("f1 score :", f1_score(y_test,y_pred))
print("confusion_matrix :", confusion_matrix(y_test,y_pred))

```

```python
from sklearn.neighbors import KNeighborsClassifier
kn= KNeighborsClassifier()
kn.fit(x_scaled,y_train)
y_pred=kn.predict(x_test_scaled)
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, confusion_matrix
print("KNN classification")
print("precison :",precision_score(y_test,y_pred))
print("accuracy :",accuracy_score(y_test,y_pred))
print("recall :", recall_score(y_test,y_pred))
print("f1 score :", f1_score(y_test,y_pred))
print("confusion_matrix :", confusion_matrix(y_test,y_pred))
```

```python
from sklearn.naive_bayes import GaussianNB
gn=  GaussianNB()
gn.fit(x_scaled,y_train)
y_pred=gn.predict(x_test_scaled)
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, confusion_matrix
print("naive bayes classification")
print("precison :",precision_score(y_test,y_pred))
print("accuracy :",accuracy_score(y_test,y_pred))
print("recall :", recall_score(y_test,y_pred))
print("f1 score :", f1_score(y_test,y_pred))
print("confusion_matrix :", confusion_matrix(y_test,y_pred))
```
