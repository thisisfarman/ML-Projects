```python
import pandas as pd
df=pd.read_csv("novagen_dataset.csv")
df.info()
```

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import accuracy_score, classification_report, recall_score

from sklearn.linear_model import LogisticRegression
from sklearn.neighbors import KNeighborsClassifier
from sklearn.ensemble import RandomForestClassifier, GradientBoostingClassifier, VotingClassifier
```

```python
X=df.drop("Target",axis=1)
Y=df["Target"]
```

```python
x_train,x_test,y_train,y_test=train_test_split(X,Y,test_size=0.2,random_state=42,stratify=Y)
```

```python
sc=StandardScaler()
xstrain=sc.fit_transform(x_train)
xstest=sc.transform(x_test)
```

```python
lr=LogisticRegression(max_iter=100)
lr.fit(xstrain,y_train)
y_pred=lr.predict(xstest)
print("logistic regression")
print("aacuracy score :",accuracy_score(y_test,y_pred))
print("recall score :",recall_score(y_test,y_pred))
```

```python
kn=KNeighborsClassifier(n_neighbors=5)
kn.fit(xstrain,y_train)
y_pred=kn.predict(xstest)
print("KNN")
print("aacuracy score :",accuracy_score(y_test,y_pred))
print("recall score :",recall_score(y_test,y_pred))
```

```python
rdc=RandomForestClassifier(
    n_estimators=200,
    max_depth=None,
    random_state=42
)
rdc.fit(xstrain,y_train)
y_pred=rdc.predict(xstest)
print("random forest classifier")
print("aacuracy score :",accuracy_score(y_test,y_pred))
print("recall score :",recall_score(y_test,y_pred))
```

```python
gbc=GradientBoostingClassifier(
    n_estimators=150,
    learning_rate=0.1,
    max_depth=4,
    random_state=42
)
gbc.fit(xstrain,y_train)
y_pred=gbc.predict(xstest)
print("GradientBoostingClassifier")
print("aacuracy score :",accuracy_score(y_test,y_pred))
print("recall score :",recall_score(y_test,y_pred))
```

```python
VC=VotingClassifier(
    estimators=[
       ("lr", LogisticRegression(max_iter=1000, solver="liblinear")),
        ("knn", KNeighborsClassifier(n_neighbors=5)),
        ("rf", RandomForestClassifier(n_estimators=200, random_state=42))
        
    ],
    voting="soft"
)
```

```python
VC.fit(xstrain,y_train)
y_pred=VC.predict(xstest)
print("VotingClassifier")
print("aacuracy score :",accuracy_score(y_test,y_pred))
print("recall score :",recall_score(y_test,y_pred))
```
